```python
#tipe data Boolean
print(True)

#tipe data String
print("string dengan menggunakan tanda kutip dua")
print('ini string menggunakan tanda kutip satu')
#tipe data string dengan menjelaskan spesial karakter atau escape sequences
print('ini adalah tanda single quote (\')')
print("ini adalah tanda double quote (\")")
print("ini adalah tanda slash (\\)")
print("Algoritma\nPemrograman") #menggunakan \n
print("Algoritma\bPemrograman") #menggunakan \b
print("Algoritma\tPemrograman") #menggunakan \t
print("Algoritma\rPemrograman") #menggunakan \b

#tipe data Integer
print(20)

#tipe data Float
print(3.14)
print(.2)
print(4.2e-3)

#tipe data Binary
print(0b10)

#tipe data octal
print(0o10)

#tipe data Hexadecimal
print('tipe data heksa desimal',0x10)

#tipe data Complex
print(5j)

#tipe data List
print([1,2,3,4,5])
print(["satu", "dua", "tiga"])

#tipe data Tuple
print((1,2,3,4,5))
print(("satu", "dua", "tiga"))

#tipe data Dictionary
print({"nama":"Budi", 'umur':20})
#tipe data Dictionary dimasukan ke dalam variabel biodata
biodata = {"nama":"Budi", 'umur':21} #proses inisialisasi variabel biodata
print(biodata) #proses pencetakan variabel biodata yang berisi tipe data Dictionary
type(biodata) #fungsi untuk mengecek jenis tipe data. akan tampil <class 'dict'> yang berarti dict adalah tipe data dictionary
```

```python
print("Praktikum Pemrograman Artificial Intelligence - Variable dan Tipe Data")

narkoba = False
print(narkoba)
print(type(narkoba))

teks = "Ini adalah tulisan berupa String"
print(teks)

angka = 100
print(angka)

desimal = 0.001
print(desimal)

heksadesimal = 0x01
print("bilangan desimal dari 0x01 adalah", heksadesimal)

biner = 0b1010
print(biner)

kompleks = 2 + 6j
print(type(kompleks))
print(kompleks)
print(type(kompleks))

daftar_angka = [96, 97, 98, 99, 100]
print(daftar_angka)

daftar_teks = ["seratus", "dua ratus", "tiga ratus"]
print(daftar_teks)

mahasiswa = {"nama": "Ani", "umur": 19}
print(mahasiswa)

kalimat = "This string contains a single quote (') character."
print(kalimat)
```
